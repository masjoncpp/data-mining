## 1. Konsep Dasar Data Mining & Kerangka Kerja CRISP-DM

**Data Mining** merupakan proses analisis terkomputasi untuk menemukan pola, tren, korelasi, dan anomali tersembunyi dalam kumpulan data berskala besar (*Big Data*).

### 🔄 Metodologi CRISP-DM:
Kerangka kerja ini membagi siklus proyek penambangan data ke dalam 6 fase berulang (*iterative*):
1. **Business Understanding**: Memahami sasaran dan kebutuhan bisnis, serta mendefinisikan problem data mining secara jelas.
2. **Data Understanding**: Mengumpulkan data mentah, memeriksa struktur, mengidentifikasi kualitas data, dan menemukan wawasan awal.
3. **Data Preparation**: Membersihkan data, menangani *missing values*, menyaring *outliers*, rekayasa fitur (*feature engineering*), dan transformasi variabel.
4. **Modeling**: Memilih dan melatih algoritma machine learning yang sesuai (klasifikasi, regresi, atau klasterisasi).
5. **Evaluation**: Mengevaluasi performa model terhadap metrik teknis dan tujuan bisnis awal.
6. **Deployment**: Mengimplementasikan model ke lingkungan produksi atau menyajikan hasil analisis untuk pengambilan keputusan strategis.

## 3. Fondasi Komputasi Saintifik: NumPy
NumPy menyediakan objek array N-dimensi (`ndarray`) berkecepatan tinggi yang menjadi dasar komputasi vektorisasi dalam Python.


```python
# 1. Pembuatan Array 1D dan 2D
array_1d = np.array([12.5, 14.8, 18.2, 22.0, 31.4, 45.0])
matrix_2d = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print("Array 1D:", array_1d)
print(f"Dimensi Matrix 2D: {matrix_2d.shape}")
print(f"Tipe Data: {array_1d.dtype}")

# 2. Operasi Statistik dan Vektorisasi
mean_val = np.mean(array_1d)
std_val = np.std(array_1d)
z_scores = (array_1d - mean_val) / std_val

print()
print(f"Rata-rata (Mean) : {mean_val:.2f}")
print(f"Standar Deviasi  : {std_val:.2f}")
print("Z-Scores (Hasil Standardisasi):", np.round(z_scores, 2))
```

## 4. Manipulasi Data Tabular: Pandas DataFrame
Pandas menyediakan struktur data berkemampuan tinggi berupa `DataFrame` (tabel 2 dimensi) dan `Series` (1 dimensi berlabel).


```python
# Membuat DataFrame simulasi mahasiswa & performa akademik
data_mahasiswa = {
    'NIM': ['DM001', 'DM002', 'DM003', 'DM004', 'DM005', 'DM006', 'DM007', 'DM008'],
    'Nama': ['Andi', 'Budi', 'Citra', 'Dewi', 'Eko', 'Fani', 'Gilang', 'Hani'],
    'Kehadiran_Pct': [95, 88, 70, 98, 85, 92, 65, 100],
    'Nilai_Tugas': [85, 78, 65, 92, 80, 88, 60, 95],
    'Nilai_UTS': [82, 75, 68, 90, 78, 84, 55, 96],
    'Nilai_UAS': [88, 80, 64, 94, 82, 89, 58, 98],
    'Status_Aktif': [True, True, True, True, True, True, False, True]
}

df_mhs = pd.DataFrame(data_mahasiswa)
print("=== 5 Baris Pertama Data Mahasiswa ===")
df_mhs.head()
```

### 4.1. Memeriksa Struktur dan Tipe Data
Langkah fundamental dalam fase *Data Understanding* adalah memeriksa tipe data dan nilai yang hilang (*missing values*).


```python
# Memeriksa informasi ringkas DataFrame
print("--- Ringkasan Struktur Data ---")
df_mhs.info()

print()
print("--- Ringkasan Statistik Deskriptif ---")
df_mhs.describe().round(2)
```

### 4.2. Penambahan Fitur (Feature Engineering) & Slicing
Menghitung Nilai Akhir dengan bobot:
- 20% Tugas
- 30% UTS
- 50% UAS

Serta menentukan kategori predikat akademik.


```python
# Menghitung Nilai Akhir dengan Vektorisasi Pandas
df_mhs['Nilai_Akhir'] = (
    0.20 * df_mhs['Nilai_Tugas'] + 
    0.30 * df_mhs['Nilai_UTS'] + 
    0.50 * df_mhs['Nilai_UAS']
).round(2)

# Menentukan Predikat Kelulusan
def beri_predikat(nilai):
    if nilai >= 85:
        return 'A (Sangat Baik)'
    elif nilai >= 75:
        return 'B (Baik)'
    elif nilai >= 65:
        return 'C (Cukup)'
    else:
        return 'D (Kurang)'

df_mhs['Predikat'] = df_mhs['Nilai_Akhir'].apply(beri_predikat)

# Menampilkan Mahasiswa dengan Nilai Akhir di atas 80
filter_unggul = df_mhs[df_mhs['Nilai_Akhir'] >= 80]
print("--- Mahasiswa dengan Nilai Akhir >= 80 ---")
filter_unggul[['NIM', 'Nama', 'Nilai_Akhir', 'Predikat']]
```

## 5. Visualisasi Eksplorasi Awal
Menampilkan visualisasi awal performa nilai mahasiswa menggunakan Seaborn dan Matplotlib.


```python
# Visualisasi Perbandingan Nilai Akhir
plt.figure(figsize=(10, 4.5))
ax = sns.barplot(x='Nama', y='Nilai_Akhir', data=df_mhs, hue='Nama', palette='Blues_d', legend=False)

# Menambahkan garis batas kelulusan
plt.axhline(75, color='red', linestyle='--', linewidth=1.5, label='Batas Kelulusan (75)')

# Kustomisasi judul dan label
plt.title('Distribusi Nilai Akhir Mahasiswa (Minggu 1)', fontsize=13, fontweight='bold', pad=12)
plt.xlabel('Nama Mahasiswa', fontsize=11)
plt.ylabel('Nilai Akhir', fontsize=11)
plt.ylim(0, 105)
plt.legend(loc='lower right')

# Menampilkan label angka di atas setiap bar
for p in ax.patches:
    ax.annotate(f"{p.get_height():.1f}", 
                (p.get_x() + p.get_width() / 2., p.get_height()),
                ha='center', va='bottom', fontsize=9, xytext=(0, 3), 
                textcoords='offset points')

plt.tight_layout()
plt.show()
```
